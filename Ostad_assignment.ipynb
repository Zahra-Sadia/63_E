```python

class Person:
    def __init__(self, name):
        self.name = name

    def show_role(self):
        print("I am a person.")
class Student(Person):
    def show_role(self):
        print(self.name, "is a student.")

class GraduateStudent(Student):
    def show_role(self):
        print(self.name, "is a graduate student.")

class Teacher(Person):
    def show_role(self):
        print(self.name, "is a teacher.")

s1 = Student("John")
s1.show_role()
s2 = GraduateStudent("Alice")
s2.show_role()
t1 = Teacher("Mr. Smith")
t1.show_role()
```

```python

class Student:
    def __init__(self, name, marks):
        self.__name = name
        self.__marks = marks
    def get_name(self):
        return self.__name
    def set_name(self, name):
        self.__name = name
    def get_marks(self):
        return self.__marks
    def set_marks(self, marks):
        self.__marks = marks
    def display(self, message="Student Details"):
        print(message)
        print("Name:", self.__name)
        print("Marks:", self.__marks)
class Sports:
    def sports_activity(self):
        print("Student participates in sports.")


class Result(Student, Sports):
    def show_result(self):
        print("Result is displayed.")
student = Result("John", 85)

print("Name:", student.get_name())
print("Marks:", student.get_marks())

student.set_marks(90)

print("\nAfter updating marks:")
student.display()
print()
student.display("Updated Student Details")
student.sports_activity()
student.show_result()
```
